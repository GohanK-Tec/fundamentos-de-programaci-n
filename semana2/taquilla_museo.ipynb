**Explicacion de mi programa:** Resuelve el problema que tiene un museo a la hora de cobrar porque se equivocan mucho cuando hacen los descuentos y cobran mal entonces mi programa va a solucionar ese tema y va a cobrar de la manera correcta para que no existan accidentes ni reclamo de parte de clientes ni de perdida de dinero para el museo

```python
precionino = 0.0  
preciojoven = 30.0    
precioadulto = 45.0  

desc_mayor = 0.12      
descuento10 = 0.10   

total_visitantes = 0
total_descuentos = 0.0
total_recaudado = 0.0

print("=== taquilla del museo ===")
print("Puedes escribr 'fin' en cualquier momento para scerrar la taquilla")

while True:
    entrada = input("edad del visitante (o 'fin' para cerrar la taquilla): ")
    
    if entrada == "fin":
        break
        
    edad = int(entrada)
    
    if edad < 0 or edad > 110:
        print("La edad no es valida (debe ser entre 0 y 110) no se cobrará el boleto, por favor pon una edad valida")
        continue
        
    if edad < 3:
        preciobase = precionino
        tipodevisitantes = "menor de 3 años"
        descuento = 0.0
    elif edad <= 17:
        preciobase = preciojoven
        tipodevisitantes = "de 3 a 17 años"
        descuento = 0.0
    else:
        preciobase = precioadulto
        tipodevisitantes = "adulto"
        
        print("  Elige un tipo de visitante: 1) Estudiante  2) Profesor  3) Ninguno")
        opcion_tipo = int(input("  elije una de las opciones que salen en el menú: "))
        
        if edad >= 60:
            descuento = preciobase * desc_mayor
            tipodevisitantes = "adulto mayor (el desscuento es automatico)"
        elif opcion_tipo == 1:
            descuento = preciobase * descuento10
            tipodevisitantes = "estudiante"
        elif opcion_tipo == 2:
            descuento = preciobase * descuento10
            tipodevisitantes = "profesor"
        else:
            descuento = 0.0
            tipodevisitantes = "ninguno"

    totaldepagar = preciobase - descuento
    
    total_visitantes += 1
    total_descuentos += descuento
    total_recaudado += totaldepagar
    
    print(f" detalles del visitante {total_visitantes} || edad del visitante: {edad} || tipo de visitante: {tipodevisitantes} || precio base del boleto: ${preciobase:.2f} || descuento para el visitante: ${descuento:.2f} || total de pagar: ${totaldepagar:.2f}\n")

print()
print("=== resumen de la taquilla ===")
print(f"visitantes cobrados: {total_visitantes}")
print(f"total en descuentos: ${total_descuentos:.2f}")
print(f"total a pagar: ${total_recaudado:.2f}")
```
