EJERCICIO 1

```python
semana = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
distancias = (120, 85, 0, 210, 95, 0, 150)

def calcular_ruta(lista_km):
    suma_total = 0
    conteo_dias = 0
    mayor = -1
    pos_mayor = 0
    
    for pos in range(len(lista_km)):
        valor = lista_km[pos]
        suma_total += valor
        if valor > 0:
            conteo_dias += 1
        if valor > mayor:
            mayor = valor
            pos_mayor = pos
            
    media = suma_total / conteo_dias
    return suma_total, conteo_dias, round(media, 1), pos_mayor

t, c, p, idx = calcular_ruta(distancias)
print("Total:", t, "km")
print("Días con rodada:", c)
print("Promedio por día rodado:", p, "km")
print("Día más largo:", semana[idx], f"({distancias[idx]} km)")
```

EJERCICIO 2 

```python
resultados = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

def revisar_votos(datos):
    total_v = 0
    lider = ""
    top_votos = -1
    
    for grupo, cant in datos:
        total_v += cant
        if cant > top_votos:
            top_votos = cant
            lider = grupo
            
    media = total_v / len(datos)
    print("Ganador:", lider, "con", top_votos, "votos")
    print("Promedio:", media)
    print("Arriba del promedio:")
    
    for grupo, cant in datos:
        if cant > media:
            print("-", grupo, f"({cant})")
            
    return media

revisar_votos(resultados)
```

EJERCICIO 3 

```python
puntos = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

def ver_tramos(ruta):
    mayor_km = -1
    ruta_top = ""
    
    for idx in range(len(ruta) - 1):
        ciudad1, km1 = ruta[idx]
        ciudad2, km2 = ruta[idx + 1]
        dif = km2 - km1
        print(f"{ciudad1} → {ciudad2}: {dif} km")
        
        if dif > mayor_km:
            mayor_km = dif
            ruta_top = f"{ciudad1} → {ciudad2}"
            
    print(f"Tramo más largo: {ruta_top} ({mayor_km} km)")

ver_tramos(puntos)
```

```python
almacen = {"bujía": 10, "cadena": 2, "balata": 6, "filtro": 4}
orden = [("bujía", 3), ("cadena", 5), ("espejo", 1), ("balata", 6), ("filtro", 1)]
sin_stock = []

for art, num in orden:
    if art not in almacen:
        print(f"{art}: no existe en el taller")
    elif almacen[art] < num:
        print(f"{art}: stock insuficiente (hay {almacen[art]})")
    else:
        almacen[art] -= num
        print(f"{art}: surtido ({num})")

for art, queda in almacen.items():
    if queda == 0:
        sin_stock.append(art)

print(almacen)
print("Agotados:", sin_stock)
```

```python
codigos = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def usar_cupon(monto, clave):
    clave_ok = clave.strip().upper()
    if clave_ok in codigos:
        porc = codigos[clave_ok]
        return monto - (monto * (porc / 100))
    else:
        print("Cupón no válido")
        return monto

lista_cod = ["  moto20 ", "GRATIS", "metal10"]
cuenta = 1200
for c in lista_cod:
    print(f"'{c}' → ${usar_cupon(cuenta, c)}")
```

EJERCICIO 20 

```python
groserias = ["tonto", "menso", "baboso"]

def filtrar_texto(texto):
    lista_p = texto.split()
    total_c = 0
    for pos in range(len(lista_p)):
        limpia = lista_p[pos].lower().replace(",", "").replace(".", "")
        if limpia in groserias:
            lista_p[pos] = "*" * len(lista_p[pos])
            total_c += 1
            
    return " ".join(lista_p), total_c

pruebas_txt = ["no seas menso y pásame el código", "Tonto el que no haga commit", "todo bien profe"]
for p in pruebas_txt:
    res_txt, cant = filtrar_texto(p)
    print(f"{res_txt}  ({cant} censuradas)")
```

```python
vehiculos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def tarifa(h_ent, h_sal):
    tiempo = h_sal - h_ent
    if tiempo <= 1:
        return 20
    return 20 + (tiempo - 1) * 15

caja = 0
for mat, ini, fin in vehiculos:
    total_p = tarifa(ini, fin)
    caja += total_p
    print(f"{mat}: {fin - ini} h → ${total_p}")

print(f"Total del día: ${caja}")
```

```python
estudiantes = ("Ana", "Luis", "Mía", "Beto")
notas_matriz = [[85, 90, 78, 92], [60, 75, 70, 68], [95, 98, 100, 97], [70, 65, 80, 75]]

def calcular_prom(lista):
    return sum(lista) / len(lista)

top_est = ""
mayor_p = -1

for idx in range(len(estudiantes)):
    p = calcular_prom(notas_matriz[idx])
    print(f"{estudiantes[idx]}: {p}")
    if p > mayor_p:
        mayor_p = p
        top_est = estudiantes[idx]

for c in range(len(notas_matriz[0])):
    valores_col = []
    for f in range(len(notas_matriz)):
        valores_col.append(notas_matriz[f][c])
    print(f"Parcial {c+1}: {calcular_prom(valores_col)}")

print(f"Mejor alumno: {top_est}")
```

```python
lista_canciones = ["Master of Puppets - Metallica - 8:35", "Holy Wars - Megadeth - 6:32",
                   "Fade to Black - Metallica - 6:57", "The Trooper - Iron Maiden - 4:10",
                   "Hangar 18 - Megadeth - 5:14", "Aces High - Iron Maiden - 4:31"]

tiempos_banda = {}

for linea in lista_canciones:
    partes = linea.split(" - ")
    banda = partes[1]
    mins, segs = partes[2].split(":")
    total_s = int(mins) * 60 + int(segs)
    
    if banda not in tiempos_banda:
        tiempos_banda[banda] = 0
    tiempos_banda[banda] += total_s

print(tiempos_banda) 
```
