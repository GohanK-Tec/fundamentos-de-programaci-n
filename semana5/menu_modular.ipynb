1. TUPLAS, parte A 

```python
def calcular_total(tup_datos):
    acumulador = 0
    for elemento in tup_datos:
        acumulador = acumulador + elemento
    return acumulador

def seccion_tuplas():
    print("\n-----------------------------------------")
    print("           1. SECCIÓN DE TUPLAS          ")
    print("-----------------------------------------")
    
    calificaciones = (9, 7, 10, 6, 8)
    print("Tupla inicial de las calificaciones:", calificaciones)
    
    
    print("El tercer elemento es:", calificaciones[2])
    

    val1 = int(input("Ingresa un número extra: "))
    val2 = int(input("Ingresa otro número extra: "))
    tupla_ampliada = calificaciones + (val1, val2)
    print("Tupla nueva combinada:", tupla_ampliada)
    
    lista_temporal = list(tupla_ampliada)
    lista_temporal.sort()
    print("Lista ordenada de menor a mayor:", lista_temporal)

    suma_total = calcular_total(tupla_ampliada)
    print("La suma de todos los elementos es:", suma_total)

seccion_tuplas()    
```

2. DICCIONARIOS, parte B 

```python
def buscar_contacto(directorio, nombre_buscado):
    if nombre_buscado in directorio:
        return directorio[nombre_buscado]
    else:
        return f"{nombre_buscado} no está registrado en la agenda, pon otro nombre"

def seccion_diccionarios():
    print("\n-----------------------------------------")
    print("        2. SECCIÓN DE DICCIONARIOS       ")
    print("-----------------------------------------")

    amigos = {
        "Sofia": "442-123-4567",
        "Brenda": "442-987-6543",
        "Tadeo": "442-555-8899"
    }
    

    nuevo_amigo = input("Escribe el nombre de quien quieras agregar: ").strip().title()
    nuevo_numero = input("Escribe tambien el número de teléfono de esa persona: ").strip()
    amigos[nuevo_amigo] = nuevo_numero
    
    print("\n[ Lista de contactos en la agenda ]")
    for amigo in amigos.keys():
        print("  ->", amigo)
        
    consultar = input("\n¿A qué contacto quieres consultar?: ").strip().title()
    respuesta = buscar_contacto(amigos, consultar)
    print(f"Resultado de búsqueda: {respuesta}")

seccion_diccionarios()
```

3. EXCEPCIONES, parte C

```python
def seccion_excepciones():
    print("\n-----------------------------------------")
    print("         3. SECCIÓN DE EXCEPCIONES       ")
    print("-----------------------------------------")
    try:
        x = int(input("Escribe el numerador: "))
        y = int(input("Escribe el denominador: "))
        
        resultado_suma = x + y
        resultado_division = x / y
        print("La suma de los dos numeros es:", resultado_suma)
        print("La división del numerador entre el denominador es:", resultado_division)
        
    except ValueError:
        print("Erroooorr, pusiste letras o dejaste el espacio vacío. Se requieren números enteros")
    except ZeroDivisionError:
        print("Erroooorr, no se puede dividir un número entre cero.")

seccion_excepciones()
```

4. STRINGS, parte D

```python
def contar_fichas_palabras(frase):
    lista_palabras = frase.split()
    return len(lista_palabras)

def seccion_strings():
    print("\n-----------------------------------------")
    print("           4. SECCIÓN DE STRINGS         ")
    print("-----------------------------------------")

    oracion = "espero sacar 100 en esta entrega, Amen"
    print("Oración original:", oracion)
    
    print("Número de caracteres (longitud):", len(oracion))
    
    print("Oración en mayúsculas:", oracion.upper())
    
    oracion_modificada = oracion.replace("bien", "excelente")
    print("Oración con palabra reemplazada:", oracion_modificada)
    
    conteo = contar_fichas_palabras(oracion)
    print("Número de palabras en la oración:", conteo)

seccion_strings()
```

5. EL MENÚ, parte E 

```python
while True:
    print("\n=========================================")
    print("          MENÚ PRINCIPAL DE OPCIONES     ")
    print("=========================================")
    print("  [1] Tuplas")
    print("  [2] Diccionarios")
    print("  [3] Excepciones")
    print("  [4] Strings")
    print("  [5] Salir del sistema")
    print("-----------------------------------------")
    
    seleccion = input("Selecciona una opción del 1-5: ").strip()
    
    if seleccion == "1":
        seccion_tuplas()
    elif seleccion == "2":
        seccion_diccionarios()
    elif seleccion == "3":
        seccion_excepciones()
    elif seleccion == "4":
        seccion_strings()
    elif seleccion == "5":
        print("Perfecto, quisiste salir del programa, byee!")
        break
    else:
        print("Opción incorrecta, trata otra vez con un número del 1 al 5")
```
