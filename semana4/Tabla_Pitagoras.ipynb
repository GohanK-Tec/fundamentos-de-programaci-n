# Introducción

En esta actividad vamos a crear una tabla de Pitágoras utilizando Python. Esta tabla nos permite ver los resultados de las multiplicaciones del 1 al 10 de una manera ordenada y fácil de consultar. El objetivo es poner en práctica algunos de los temas que hemos aprendido durante el curso, como el uso de listas, ciclos `for`, funciones y ciclos `while`.

Primero, vamos a crear una matriz de 10 filas y 10 columnas. Para llenarla, utilizaremos dos ciclos `for`, uno para recorrer los renglones y otro para las columnas. En cada espacio se guardará el resultado de multiplicar los dos números correspondientes. Después, haremos una función que nos permita mostrar toda la tabla en pantalla, procurando que los números aparezcan acomodados y sean fáciles de leer.

También vamos a crear otra función para buscar un resultado específico. Para esto, el programa le pedirá al usuario que escriba dos números: uno para el renglón y otro para la columna. Es importante que los dos valores estén entre 1 y 10, por lo que agregaremos una validación para que, si se escribe un número fuera de ese rango, el programa vuelva a pedirlo hasta que sea correcto.

Por último, vamos a probar el programa con diferentes números para revisar que los resultados sean correctos y que también funcione cuando se ingresan valores que no están permitidos. Con esta actividad podremos practicar cómo guardar información en una matriz, cómo utilizar funciones y cómo hacer que un programa sea más fácil de usar y entender.

```python
# Construcción de la tabla de Pitágoras

tabla = []

for numero_fila in range(1, 11):
    fila = []

    for numero_columna in range(1, 11):
        valor = numero_fila * numero_columna
        fila.append(valor)

    tabla.append(fila)

print("La matriz de Pitágoras se generó correctamente.")
print("Cantidad de filas:", len(tabla))
print("Columnas por fila:", len(tabla[0]))
```

# Explicación del código
Primero, se crea una lista vacía llamada tabla, que funcionará como la matriz principal donde se guardarán los resultados.

Después, se utiliza un ciclo for con range(1, 11) para recorrer los números del 1 al 10. Este ciclo representa los renglones de la tabla.

Dentro del primer ciclo se crea otra lista vacía llamada fila, que almacenará los resultados correspondientes al renglón actual.

Posteriormente, se utiliza un segundo ciclo for para recorrer las columnas del 1 al 10. En cada repetición se multiplica el número del renglón por el número de la columna.

El resultado se guarda temporalmente en la variable valor y se agrega a la lista fila mediante append().

Cuando termina el recorrido de las columnas, se utiliza tabla.append(fila) para incorporar la fila completa a la matriz principal.

Finalmente, se utilizan len() y print() para comprobar que la matriz contiene diez filas y que cada fila tiene diez columnas.

De esta manera, se construye automáticamente una matriz de 10 × 10 con los productos de la tabla de Pitágoras.

## Parte B. Función para imprimir la tabla

En esta sección se define una función llamada `imprimir_tabla(tabla)`, cuya finalidad es mostrar todos los elementos de la matriz de forma ordenada.

La función recibe la matriz como parámetro y utiliza dos ciclos `for` para recorrer sus filas y columnas. Para separar los números se utiliza una tabulación, evitando que aparezcan los corchetes y las comas de las listas.

Esta función solamente presenta la información en pantalla, por lo que no necesita devolver ningún valor.

```python
def imprimir_tabla(tabla):
    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()

imprimir_tabla(tabla)
```

## Explicación del código
Primero, se define la función imprimir_tabla(tabla), que recibe la matriz mediante el parámetro tabla.

Dentro de la función se utiliza un ciclo for para recorrer cada una de las filas almacenadas en la matriz.

Después, se utiliza otro ciclo for para acceder individualmente a los valores que contiene cada fila.

La instrucción print(valor, end="\t") permite mostrar cada número seguido de una tabulación, lo que ayuda a mantener los resultados separados y organizados.

Al finalizar cada fila, se utiliza print() sin argumentos para generar un salto de línea y continuar mostrando la siguiente fila.

La función no utiliza return porque su único propósito es imprimir la información y no regresar un resultado.

Finalmente, se llama a imprimir_tabla(tabla) para visualizar la matriz completa.

## Parte C. Función para consultar un producto

En esta parte se desarrolla la función `consultar_producto(tabla, renglon, columna)`, que permite recuperar un resultado específico de la matriz.

La función recibe tres parámetros: la matriz, el renglón y la columna que se desean consultar.

Para obtener el resultado se utilizan los índices de la lista mediante la expresión `tabla[renglon - 1][columna - 1]`.

Se resta uno a los valores porque Python comienza a contar las posiciones de las listas desde cero. La función utiliza `return` para entregar el resultado almacenado.

Es importante mencionar que esta función no realiza ninguna multiplicación, ya que únicamente consulta el valor que previamente fue guardado en la matriz.

```python
def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]
```

## Explicación del código
Se define la función consultar_producto(), que recibe la matriz y dos valores que indican la ubicación del producto que se desea recuperar.

La instrucción return tabla[renglon - 1][columna - 1] permite acceder directamente al elemento almacenado en la posición correspondiente.

Se resta uno al renglón y a la columna porque los índices de Python comienzan desde cero. Por ejemplo, para consultar el producto de 7 por 8, se accede a la posición tabla[6][7].

El resultado se devuelve mediante return, permitiendo que otra parte del programa pueda utilizarlo.

En esta función no se utiliza el operador *, ya que el propósito es consultar la matriz y no calcular nuevamente el producto.

## Parte D. Programa principal y validación de datos

En esta sección se desarrolla el programa principal, donde el usuario puede ingresar los dos factores que desea consultar.

Para recibir los datos se utilizan las instrucciones `input()` e `int()`, que permiten capturar los valores y convertirlos en números enteros.

Posteriormente, se utilizan ciclos `while` para verificar que ambos factores estén dentro del intervalo del 1 al 10.

Cuando un valor se encuentra fuera del rango permitido, el programa muestra un mensaje de advertencia y vuelve a solicitar el dato.

Una vez que los dos factores son válidos, se llama a la función `consultar_producto()` y se guarda el resultado que devuelve en una variable.

Finalmente, se muestra en pantalla la operación consultada y el producto obtenido de la matriz.

```python
print("CONSULTA DE LA TABLA DE PITÁGORAS")

factor_fila = int(input("Factor del renglón (1 a 10): "))

while factor_fila < 1 or factor_fila > 10:
    print("Dato no válido. El renglón debe estar entre 1 y 10.")
    factor_fila = int(input("Vuelve a escribir el factor del renglón: "))

factor_columna = int(input("Factor de la columna (1 a 10): "))

while factor_columna < 1 or factor_columna > 10:
    print("Dato no válido. La columna debe estar entre 1 y 10.")
    factor_columna = int(input("Vuelve a escribir el factor de la columna: "))

resultado = consultar_producto(tabla, factor_fila, factor_columna)

print("Resultado de la consulta:")
print(f"{factor_fila} x {factor_columna} = {resultado}")
```

## Explicación del código
Primero, se muestra el título del programa mediante print().

Después, se solicita el primer factor utilizando input() y se convierte a entero con int(). El dato se almacena en la variable factor_fila.

A continuación, se utiliza un ciclo while para comprobar que el número ingresado no sea menor que 1 ni mayor que 10.

Si el usuario introduce un valor fuera del intervalo permitido, se muestra un mensaje indicando que el dato no es válido y se vuelve a solicitar el número.

Se repite el mismo procedimiento para el segundo factor, que se almacena en factor_columna.

Esta validación es necesaria porque Python permite utilizar índices negativos para acceder a elementos de una lista. Por ejemplo, si se introduce cero y se utiliza renglon - 1, se obtiene el índice -1, que corresponde al último elemento y podría generar un resultado incorrecto.

Cuando ambos factores cumplen con las condiciones, se llama a consultar_producto(tabla, factor_fila, factor_columna).

El valor devuelto por la función se almacena en la variable resultado.

Finalmente, se utiliza print() para mostrar los factores ingresados y el resultado recuperado de la matriz.

```python
# Prueba adicional de validación de datos

print("PRUEBA DE VALIDACIÓN DE LA TABLA DE PITÁGORAS")

factor_fila = int(input("Factor del renglón (1 a 10): "))

while factor_fila < 1 or factor_fila > 10:
    print("Dato no válido. El renglón debe estar entre 1 y 10.")
    factor_fila = int(input("Vuelve a escribir el factor del renglón: "))

factor_columna = int(input("Factor de la columna (1 a 10): "))

while factor_columna < 1 or factor_columna > 10:
    print("Dato no válido. La columna debe estar entre 1 y 10.")
    factor_columna = int(input("Vuelve a escribir el factor de la columna: "))

resultado = consultar_producto(tabla, factor_fila, factor_columna)

print("Resultado de la consulta:")
print(f"{factor_fila} x {factor_columna} = {resultado}")
```

```python
print("CASOS DE PRUEBA DE LA TABLA DE PITÁGORAS")

# Caso 1: factores intermedios
print("\nCaso 1: 7 x 8")
print("Resultado obtenido:", consultar_producto(tabla, 7, 8))

# Caso 2: valores mínimos
print("\nCaso 2: 1 x 1")
print("Resultado obtenido:", consultar_producto(tabla, 1, 1))

# Caso 3: valores máximos
print("\nCaso 3: 10 x 10")
print("Resultado obtenido:", consultar_producto(tabla, 10, 10))

# Caso 4: validación de valores fuera del rango
print("\nCaso 4: 0 x 8")
factor_fila = 0
factor_columna = 8

while factor_fila < 1 or factor_fila > 10:
    print("Dato no válido. El renglón debe estar entre 1 y 10.")
    factor_fila = 7

resultado = consultar_producto(tabla, factor_fila, factor_columna)
print("Resultado obtenido después de corregir:", resultado)
```
